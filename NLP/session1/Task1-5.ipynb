```python
#task -5 

def is_spam(email_text):
    email_lower = email_text.lower()
    if "win" in email_lower or "free" in email_lower:
        return True 
    else:
        return False

email1 = "Congratulations! You win a free iPhone today!"
print(is_spam(email1))


email2 = "Hey, are we meeting tomorrow for lunch?"
print(is_spam(email2))            
```
