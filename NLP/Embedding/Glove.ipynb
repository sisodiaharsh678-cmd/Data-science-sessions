GloVe = Global Vectors
Made by Stanford (2014)

Word2Vec → learns from LOCAL context
           (nearby words only)

GloVe    → learns from GLOBAL statistics
           (how often words appear together
            across ENTIRE dataset)

```python
import gensim.downloader as api 

model = api.load("glove-wiki-gigaword-100")
print("GloVe loaded!")
print("Vocab size:", len(model))

print("\nSimilar to 'king':")
for word, score in model.most_similar("king", topn=5):
    print(f"  {word} → {score:.2f}")


print("\nSimilarity scores:")
print(f"  king  vs queen  → {model.similarity('king',  'queen'):.2f}")
print(f"  cat   vs dog    → {model.similarity('cat',   'dog'):.2f}")
print(f"  pizza vs coding → {model.similarity('pizza', 'coding'):.2f}")
```
