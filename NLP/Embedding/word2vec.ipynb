```python

from gensim.models import Word2Vec

sentences = [
    ["the", "king", "rules", "the", "kingdom"],
    ["the", "king", "sits", "on", "throne"],
    ["the", "king", "wears", "a", "crown"],
    ["the", "king", "is", "powerful", "leader"],
    ["the", "queen", "rules", "the", "palace"],
    ["the", "queen", "sits", "on", "throne"],
    ["the", "queen", "wears", "a", "crown"],
    ["the", "queen", "is", "powerful", "leader"],
    ["the", "man", "is", "strong", "and", "brave"],
    ["the", "man", "works", "hard", "every", "day"],
    ["the", "woman", "is", "strong", "and", "brave"],
    ["the", "woman", "works", "hard", "every", "day"],
    ["the", "prince", "will", "become", "king"],
    ["the", "princess", "will", "become", "queen"],
    ["kingdom", "is", "ruled", "by", "king"],
    ["palace", "is", "home", "of", "queen"],
    ["crown", "belongs", "to", "king", "and", "queen"],
    ["throne", "is", "seat", "of", "king", "and", "queen"],
]

model = Word2Vec(
    sentences,
    vector_size = 50 ,
    window = 3 , 
    min_count = 1 ,
    epochs = 50,
    
)

vec = model.wv["king"]
print("Shape:", vec.shape)

print("Vector:", vec[:5].round(3))

#simiar words 
simliar = model.wv.most_similar("king" , topn=3)
print("similar to king")
for word , score in simliar:
    print(f"  {word} -{score:.2f}")
```

```python
import gensim.downloader as api

print("Downloading GloVe... (one time only)")
model = api.load("glove-wiki-gigaword-100")


print("\nSimilar to 'king':")
for word, score in model.most_similar("king", topn=5):
    print(f"  {word} → {score:.2f}")


print("\nSimilarity scores:")
print(f"  king  vs queen  → {model.similarity('king',  'queen'):.2f}")
print(f"  cat   vs dog    → {model.similarity('cat',   'dog'):.2f}")
print(f"  pizza vs coding → {model.similarity('pizza', 'coding'):.2f}")
```
