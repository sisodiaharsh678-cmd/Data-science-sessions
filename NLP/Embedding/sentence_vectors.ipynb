 Word2Vec / GloVe → vector per WORD
ML Model needs   → vector per SENTENCE

Problem:
"I love cricket" → 3 separate word vectors
Model needs      → 1 single vector

Solution:
Average all word vectors → 1 sentence vector

"I love cricket"

vector("I")       = [0.1,  0.5,  0.3]
vector("love")    = [0.8,  0.2,  0.9]
vector("cricket") = [0.4,  0.7,  0.1]

Average:
= (0.1+0.8+0.4)/3, (0.5+0.2+0.7)/3, (0.3+0.9+0.1)/3
= [0.43, 0.46, 0.43]

→ 1 vector for whole sentence ✅

```python
#Word Analogies

import gensim.downloader as api

model = api.load("glove-wiki-gigaword-100")

# king - man + woman = ?
result = model.most_similar(
    positive=["king", "woman"],  # add these
    negative=["man"],            # subtract this
    topn=3
)

print("king - man + woman :")
for word, score in result:
    print(f"  {word} → {score:.2f}")
```
