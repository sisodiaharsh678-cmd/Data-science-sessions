Task 1 


SCENARIO
You are building an automated customer support chatbot for a food delivery platform
that handles queries like 'Where is my order?', 'I want to cancel my order', and 'My food
arrived cold and late'.

Question: Identify two NLP techniques the chatbot must use to correctly understand and route
these customer messages. Explain why each technique is necessary and describe what would
happen to chatbot accuracy if either technique were removed.

Answer : - Two essential NLP techniques are Intent Classification and Named Entity Recognition (NER).
Intent Classification determines the purpose of the message (status inquiry, cancellation, complaint) so the chatbot can route it to the correct handling flow. Without it, the chatbot cannot distinguish between fundamentally different request types and would either respond generically or rely on brittle keyword rules that fail on natural phrasing variation — severely degrading routing accuracy.

Named Entity Recognition extracts specific details from the message — order ID, item, date/time — needed to act on the request. Without NER, the chatbot would correctly identify what the customer wants but not which order it concerns, forcing repetitive clarifying questions and increasing the risk of misapplied actions (e.g. canceling the wrong order).

Together, Intent Classification answers "what does the customer want," and NER answers "what specifically does it apply to" — both are required for the chatbot to both route and resolve a query correctly, not just one or the other

Task 2 

SCENARIO
You are processing 50,000 raw customer reviews scraped from a food delivery app. The
reviews contain mixed letter cases, special characters, emojis, numbers, and common
filler words such as 'the', 'is', and 'a'.

Question: Describe the text preprocessing pipeline you would apply to this dataset before
feeding it to any NLP model. Justify the order in which you would apply each step, and explain
specifically why skipping stopword removal could hurt downstream model performance.

The preprocessing pipeline should follow this order: (1) remove or convert emojis/special characters, (2) remove numbers, (3) lowercase the text, (4) tokenize, (5) remove stopwords, (6) apply stemming or lemmatization, (7) vectorize (TF-IDF or embeddings). This order matters because each step assumes the text produced by the previous step is already clean — for example, tokenization must happen after junk-character removal, or it produces broken tokens; stopword removal and stemming must happen after tokenization, since they operate on individual words, not raw text; and stopword removal should precede stemming so we don't waste computation normalizing words we're about to discard.


Skipping stopword removal specifically hurts performance in frequency-based models (BoW, TF-IDF) because stopwords appear at high frequency across nearly all reviews regardless of sentiment, diluting the relative importance of genuinely discriminative words like "cold," "late," or "delicious." This increases feature-space size without adding useful signal, slowing training and potentially causing the model to latch onto irrelevant patterns instead of real sentiment indicators. (Note: this specific concern is less relevant for transformer-based models, which can extract useful signal from stopwords via attention.)

Task 3 

You are building a food item search feature that must rank menu descriptions by
relevance to a customer's typed query, for example 'spicy chicken without onion'.

Question: Compare Bag of Words and TF-IDF as text representation methods for this ranking
task. Which would you choose and why? Identify one limitation that your chosen method still has
even after it is applied.

TF-IDF is the better choice over Bag of Words for this ranking task. BoW simply counts word occurrences, so it can't distinguish genuinely relevant, distinctive words (like "spicy," "chicken") from common filler words that appear across nearly all menu descriptions ("served," "with") — this often causes longer, more generic descriptions to outrank more relevant, specific ones. TF-IDF fixes this by weighting words based on how rare and distinctive they are across the whole menu, so terms that actually characterize a dish get scored higher than filler text, producing better relevance ranking.

Task 4 

You are training a Naive Bayes classifier to automatically label incoming food delivery
support tickets as 'Delivery Issue', 'Food Quality Issue', or 'App Issue' using a dataset of
10,000 historical labelled tickets.

Question: Explain the core probabilistic assumption that Naive Bayes makes about its input
features. Describe a realistic food delivery example where this assumption is violated, and
explain how that violation would affect the model's classification accuracy.

Naive Bayes assumes conditional independence between features given the class — it treats every word's contribution to a class probability as if it were unrelated to every other word in the same ticket. In this scenario, that assumption breaks down with words like "cold" and "late," which frequently co-occur because a delayed delivery is often the actual cause of cold food — they aren't independent, they're two symptoms of one event.

This violation causes Naive Bayes to effectively double-count overlapping evidence, since it multiplies each word's individual probability as if they were separate signals. The practical effect is overconfident, sometimes miscalibrated predictions, and increased risk of misclassification on borderline tickets where correlated words also appear (even weakly) in a different category — though Naive Bayes often still performs reasonably well in practice, since classification only requires the correct class to score highest, not perfectly accurate probabilities.

Task 5 

You are comparing a Logistic Regression model and a Naive Bayes model for classifying
food delivery reviews as Positive, Neutral, or Negative. Your dataset is heavily
imbalanced: 80% of reviews are Positive, 15% Neutral, and 5% Negative.

Question: Describe two evaluation metrics you would prioritise over accuracy when comparing
these models on this imbalanced dataset. Explain one technique you would apply to address the
class imbalance before training.

Two metrics to prioritize over accuracy: (1) Precision, Recall, and F1-score calculated per class — especially Recall on the Negative class, since missing genuine complaints is more costly than minor confusion between Positive and Neutral; and (2) Macro-averaged F1-score, which weighs all three classes equally rather than letting the 80% Positive class dominate the score and hide poor minority-class performance.

To address the imbalance before training, I would apply SMOTE (Synthetic Minority Oversampling Technique) to generate synthetic examples for the Neutral and Negative classes, giving the model more balanced exposure during training without discarding majority-class data the way undersampling would. A simpler alternative is class weighting (e.g., class_weight='balanced'), which penalizes minority-class misclassification more heavily without altering the dataset itself.
