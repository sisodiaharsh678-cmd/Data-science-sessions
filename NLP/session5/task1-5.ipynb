Task 2 — Limitation of BoW/TF-IDF on WhatsApp messages
Explanation: This is a conceptual question  no code needed, just a clear written answer.


"Traditional models like Bag of Words or TF-IDF treat each word independently and ignore word order and context, so they can't capture real meaning. For example, in a WhatsApp message like 'I am not free today, maybe tomorrow,' these models would just count words like 'not,' 'free,' and 'tomorrow' separately, without understanding that 'not free' actually negates 'free' — leading to a completely wrong interpretation of the sentence's real meaning. Deep learning models (like transformers) solve this by understanding words in relation to their surrounding context, not in isolation."


```python
from transformers import pipeline 

sentiment_pipeline = pipeline("sentiment-analysis")

reviews = [
    "Food was delicious and delivery was super fast!",
    "Order arrived late and the food was cold, very disappointed.",
    "Amazing packaging and great taste, will order again."
]

for review in reviews:
    result = sentiment_pipeline(review)
    print(f"Review: {review}\nSentiment: {result}\n")
```

Model chosen: RoBERTa (Robustly Optimized BERT Pretraining Approach)


"RoBERTa is trained on significantly more data and for longer than BERT, and removes BERT's Next Sentence Prediction training objective, focusing purely on masked language modeling. These changes generally give RoBERTa better performance than BERT on many text classification benchmarks, despite having a similar underlying architecture."
