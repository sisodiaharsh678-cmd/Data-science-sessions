The output layer should use Softmax, not Sigmoid.
Sigmoid scores each of the three output neurons independently, so the values don't sum to 1 and don't represent a real probability distribution — the model could output high probabilities for multiple tiers at once (e.g. 0.8 for Priority and 0.75 for Express), which is invalid since an order belongs to exactly one tier.
Softmax normalizes all three outputs together so they always sum to exactly 1, correctly modeling mutually exclusive classes — raising the probability of one tier necessarily lowers the others.


Q5 - Dropout randomly disables (zeroes out) a fraction of neurons during each training pass, forcing the network to not rely too heavily on any single neuron and encouraging redundant, generalizable representations. Typical rate: 0.2–0.5 (0.3 is a common default for dense layers).
L2 regularization adds a penalty term to the loss function proportional to the sum of squared weights, discouraging any single weight from growing too large. Typical strength: λ = 0.001–0.01.
In practice, apply Dropout first — it's simpler to tune, works well for dense layers specifically, and tends to give a bigger overfitting reduction with less trial and error than L2.
Mechanism difference: Dropout changes the network's structure on each forward pass (randomly removing neurons); L2 changes the loss function itself (penalizing large weights), constraining the model without altering its architecture.

Q6 - A plain ANN flattens the image into a 1D vector immediately, destroying all spatial relationships between pixels — a pixel's neighbors are treated as no more related to it than a pixel on the opposite corner of the image. Every neuron in the first Dense layer connects to every pixel, so the network must learn spatial patterns separately for every possible position in the image, using a huge number of independent weights.


A CNN instead uses convolutional filters that slide across the image while preserving its 2D spatial structure. Each small filter (e.g. 3x3) shares the same weights at every position it passes over, rather than having separate weights per location.


What a convolutional filter learns that fully connected layers can't efficiently capture:
A filter learns to detect local spatial patterns — edges, corners, textures, curves — and because it's reused across the whole image, it can recognize that pattern anywhere it appears. This is called translation invariance: a filter trained to detect blur or a rounded edge will activate wherever that pattern occurs in the frame, not just in one fixed spot.


A fully connected layer can't do this efficiently — it would need entirely separate weights to learn "blur in the top-left" versus "blur in the bottom-right," massively increasing parameters and requiring far more training data. This makes CNNs both more accurate and more parameter-efficient for image tasks like your food-photo quality classifier, where quality issues can appear anywhere in the frame.
