```python
def calculate_reward(delivery_time, is_on_time, customer_rating):
    reward = 10 if is_on_time else -5
    reward += customer_rating
    print(f"time={delivery_time} on_time={is_on_time} rating={customer_rating} -> reward={reward}")
    return reward

calculate_reward(25, True, 5)
calculate_reward(48, False, 2)
calculate_reward(30, True, 3)
```

```python
def update_state(state, new_location, reward_earned):
    state["location"] = new_location
    state["orders_delivered"] += 1
    state["total_reward"] += reward_earned
    return state

agent_state = {"location": "Depot", "orders_delivered": 0, "total_reward": 0.0, "is_available": True}

for loc, r in [("Zone A", 8.5), ("Zone B", 12.0), ("Zone C", -3.0), ("Zone A", 15.5)]:
    agent_state = update_state(agent_state, loc, r)
    print(agent_state)
```

```python
reward_map = {'accept_order': 2, 'reject_order': -1, 'request_directions': 0, 'mark_delivered': 10}
episode = ['accept_order','request_directions','accept_order','reject_order','request_directions','mark_delivered']

cumulative = 0
for step, action in enumerate(episode, 1):
    r = reward_map[action]
    cumulative += r
    print(f"Step {step}: {action} | reward={r} | cumulative={cumulative}")
print("Total episode reward:", cumulative)  
```
