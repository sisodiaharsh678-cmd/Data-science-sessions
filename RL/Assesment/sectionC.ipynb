# Mini Project: Food Delivery RL Concept Simulator


```python
import random

ACTIONS = ['navigate_to_zone', 'accept_order', 'reject_order', 'request_directions', 'mark_delivered']

REWARD_MAP = {
    'navigate_to_zone': 1,
    'accept_order': 2,
    'reject_order': -1,
    'request_directions': 0,
    'mark_delivered': 10,
}

ZONES = ['Zone A', 'Zone B', 'Zone C', 'Zone D']

class DeliveryAgentSimulator:
    def __init__(self):
        # Agent's persistent state -- carries over across episodes
        self.state = {"location": "Depot", "orders_delivered": 0, "total_reward": 0.0}
        self.episodes_run = 0

    def run_episode(self, num_actions=6):
        """Simulates one episode: a fixed-length sequence of actions,
        updating state and reward after every single action."""
        episode_log = []
        episode_reward = 0

        print(f"\n--- Episode {self.episodes_run + 1} in progress ---")
        for step in range(1, num_actions + 1):
            action = random.choice(ACTIONS)
            reward = REWARD_MAP[action]

            # State updates depend on which action was taken
            if action == "navigate_to_zone":
                self.state["location"] = random.choice(ZONES)
            elif action == "mark_delivered":
                self.state["orders_delivered"] += 1

            self.state["total_reward"] += reward
            episode_reward += reward

            episode_log.append((step, action, reward))
            print(f"  Step {step}: action='{action}' | reward={reward} | "
                  f"cumulative episode reward={episode_reward}")

        self.episodes_run += 1

        # Full episode summary
        print("\n  --- Episode Summary ---")
        for step, action, reward in episode_log:
            print(f"    Step {step}: {action} (reward={reward})")
        print(f"  Total episode reward: {episode_reward}")
        print(f"  Agent state after episode: {self.state}\n")

    def view_stats(self):
        print("\n=== Agent Stats ===")
        print(f"Total episodes run:        {self.episodes_run}")
        print(f"Total reward earned:       {self.state['total_reward']}")
        avg = (self.state['total_reward'] / self.episodes_run) if self.episodes_run > 0 else 0
        print(f"Average reward per episode: {avg:.2f}")
        print(f"Total orders delivered:    {self.state['orders_delivered']}")
        print(f"Current location:          {self.state['location']}\n")


def main():
    sim = DeliveryAgentSimulator()
    while True:
        print("===== Food Delivery RL Simulator =====")
        print("1. Start New Delivery Episode")
        print("2. View Agent Stats")
        print("3. Exit")
        choice = input("Choose an option (1-3): ").strip()

        if choice == "1":
            sim.run_episode()
        elif choice == "2":
            sim.view_stats()
        elif choice == "3":
            print("\nExiting simulator. Final stats:")
            sim.view_stats()
            break
        else:
            print("Invalid option, please choose 1, 2, or 3.\n")


if __name__ == "__main__":
    main()
```
