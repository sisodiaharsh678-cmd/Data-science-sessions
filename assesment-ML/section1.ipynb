Q3. Model performs well on training but poorly on test data

it's called overfitting. 
Overfitting occurs when a model learns the training data too well — including its noise and random patterns — and fails to generalise to new unseen data.

Causes of Overfitting:

Model is too complex (too many parameters)
Dataset is too small
Too many features relative to data size
Model trained for too many iterations

How to fix it:

Use simpler model
Apply cross-validation
Use regularisation (L1/L2)
Add more training data
Use feature selection to reduce irrelevant features

q5.Purpose of the Test Set
The test set is a portion of the dataset that the model never sees during training. It is kept separate specifically to evaluate how well the model performs on completely new, unseen data.
