
1. Data Dictionary
A Data Dictionary was created to document all 35 features of the IBM HR Attrition dataset. Each feature was mapped to its data type (integer or object), statistical type (continuous, ordinal, nominal, binary or constant), and a brief description. This process helped identify four constant columns — EmployeeCount, EmployeeNumber, Over18, and StandardHours — which were later dropped as they carry no predictive value for the model.

3.Git Versioning Strategy
-A Git-based versioning strategy was implemented to track all iterations of the HR attrition model. The repository was structured with separate folders for data, notebooks, and saved models. Each development milestone was committed with a descriptive message including the model type and accuracy achieved. This approach ensures reproducibility, allows rollback to any previous version, and follows industry-standard MLOps practices. The repository was hosted on GitHub for remote backup and version history tracking.

4.End-to-End ML Architecture
The end-to-end ML architecture follows a structured eight-stage pipeline. Beginning with raw CSV data ingestion, the pipeline progresses through exploratory data analysis, preprocessing, model training, and evaluation before concluding with hyperparameter tuning and deployment. Four classification models were trained and compared, with the best model saved using joblib and deployed as an interactive Streamlit web application. This architecture ensures a reproducible, modular, and scalable workflow that follows industry-standard MLOps practices.
