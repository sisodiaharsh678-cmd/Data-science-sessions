## 2. Zomato-style Q&A chatbot
A simple canned-response bot: given a restaurant name, it looks up the cuisine in a small dictionary and returns a fixed-format sentence.

```python
RESTAURANT_DB = {
    "pizza palace": "Italian",
    "spice route": "Indian",
    "sushi central": "Japanese",
    "taco fiesta": "Mexican",
    "the noodle house": "Chinese",
    "curry leaf": "South Indian",
}

def get_restaurant_info(name: str) -> str:
    """Return a canned response for a given restaurant name."""
    key = name.strip().lower()
    if key in RESTAURANT_DB:
        cuisine = RESTAURANT_DB[key]
        return f"Yes, {name.strip()} serves {cuisine} cuisine."
    return f"Sorry, I don't have information about '{name.strip()}' yet."

# Try it out
print(get_restaurant_info("Pizza Palace"))
print(get_restaurant_info("Sushi Central"))
print(get_restaurant_info("Unknown Place"))
```

## 4. Song filter function (Copilot/ChatGPT-assisted)
**Prompt used:** \"Write a Python function that takes a list of song names and returns only the ones that start with the letter 'S', case-insensitive.\"

**Generated code:**

```python
def filter_songs_starting_with_s(songs):
    """Return only the songs whose name starts with 'S' (case-insensitive)."""
    return [song for song in songs if song.strip().upper().startswith("S")]

playlist = [
    "Shape of You",
    "Blinding Lights",
    "Sunflower",
    "Levitating",
    "Someone Like You",
    "Stay",
    "Circles",
]

filtered = filter_songs_starting_with_s(playlist)
print("Original playlist:", playlist)
print("Songs starting with 'S':", filtered)
```

## 5. Production challenges for Gen AI chatbots (BookMyShow / Myntra style apps)

**Challenge 1: Hallucination / factually wrong answers**
Chatbots can confidently state incorrect info — wrong showtimes, wrong return policy details — which erodes trust and can cause real business problems.
*Solution:* Ground responses in a Retrieval-Augmented Generation (RAG) setup pulling from the company's live database, and instruct the model to answer only from retrieved context.

**Challenge 2: Latency and cost at scale**
Serving millions of users with large LLMs is expensive and can introduce noticeable delays during high-traffic periods (release days, sale events).
*Solution:* Route simple/common queries to a smaller distilled model, cache frequent responses, and reserve the large model for complex queries.

# Traditional AI/ML vs Generative AI

## 1. Comparison table

| Aspect | Traditional AI/ML | Generative AI |
|---|---|---|
| **Data requirements** | Structured, labeled data (ratings, clicks, purchase history) | Massive unstructured data (text, images, audio) used for pretraining |
| **Output type** | A prediction, score, or classification (a rating, a category, a ranked list) | New content — text, images, code, audio generated on the fly |
| **Human interaction** | Mostly invisible — works in the background to rank/filter things for you | Often conversational/direct — you prompt it and it responds like a collaborator |
| **Adaptability** | Needs retraining on new labeled data to adapt to new patterns | Can adapt to a new task instantly via prompting, no retraining needed |
| **Example use case** | **Spotify's Discover Weekly** — scores songs from listening history and similar users | **Spotify's AI DJ** — generates natural-language commentary between songs, dynamically per user |
| **Example use case (2nd)** | **Zomato's delivery-time estimate** — regression model predicts ETA from distance/traffic/load | **Instagram's AI caption/reply suggestions** — generates new phrasing based on image and context |


## 2. Is Zomato's order-history-based recommendation Traditional AI/ML or Gen AI?

**Reasoning:** The key test for any recommendation system is whether it's *selecting/ranking from existing items* or *creating something new*. Zomato isn't writing you a new restaurant — it's scoring restaurants that already exist in its database using patterns from your order history.

**Answer:** This is traditional AI/ML. It analyzes structured data — past orders, ratings, and similar users' behavior — and uses a recommendation algorithm (e.g., collaborative filtering) to rank existing restaurants by predicted relevance. No new content is generated; the system selects and orders pre-existing options, which is a classic pattern-matching task rather than a generative one.

## 4. Flipkart customer support chatbot — Traditional vs Gen AI

**Reasoning:** The two dimensions to compare are *how the response is produced* and *how flexible it is*. Traditional approaches match the query to a known pattern and return a fixed/templated answer; Gen AI composes a new answer in context, trading predictability for flexibility.

**Traditional AI/ML approach:**
1. Intent classification — classify the query into a known category (e.g., \"refund status\") and trigger a pre-written canned response for that category.
2. Decision-tree/FAQ matching — match the query against a keyword/FAQ database and return the closest stored answer verbatim.

**Gen AI approach:**
1. Dynamic response generation — compose a fresh, context-specific response (e.g., referencing the customer's actual order number and delivery date), phrased naturally rather than picked from a template.
2. Handling novel/unstructured queries — a query that doesn't match any predefined category can still get a coherent answer, because the model reasons over the request instead of requiring an exact pattern match.
