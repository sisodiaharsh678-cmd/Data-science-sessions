1. Adobe Photoshop — "Generative Fill"
Uses diffusion-based inpainting (built on Adobe's Firefly model) to let users select a region of a photo and have the AI generate realistic, context-aware content to fill that space.
User scenario: A photographer takes a great landscape shot, but a stray tourist walks into the frame. Instead of manually cloning/patching pixels in the old-fashioned way, they select the person with the lasso tool, use Generative Fill, and the AI seamlessly reconstructs the background (grass, sky, rocks) as if the person was never there.
2. Topaz Gigapixel AI
Uses diffusion/deep-learning-based super-resolution to upscale low-resolution images into much higher-resolution versions, adding plausible fine detail rather than just stretching pixels (which is what older, non-AI upscaling methods do, and why they look blurry).
User scenario: Someone has an old family photo that was scanned at low resolution years ago. They want to print it large for a family reunion display. Gigapixel AI upscales the image several times over, intelligently filling in realistic texture and detail so the enlarged print doesn't look pixelated or blurry.
