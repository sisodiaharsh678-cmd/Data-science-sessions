Risk 1: Misinformation via realistic fake images
Text-to-image models can generate convincing but entirely fabricated photos — for example, a fake image showing a real public figure at an event that never happened, or a fabricated scene of a disaster or protest. On social media, such images can spread extremely fast and be mistaken for genuine photographs before anyone verifies them, contributing to the rapid spread of misinformation.

Mitigation: Developers can embed invisible digital watermarks into every AI-generated image (e.g., Google's SynthID), so social media platforms can automatically detect and label AI-generated content — flagging it to users before it's widely shared, rather than relying on manual fact-checking after the fact.

Risk 2: Amplification of bias and stereotypes
Since these models are trained on massive internet-scraped datasets, they can absorb and reproduce societal biases — for example, generating stereotyped depictions of certain professions, ethnicities, or genders even from neutral prompts. Because social media has enormous reach and virality, a single biased generated image shared widely can reinforce harmful stereotypes at scale, far more than a private or isolated use of the tool would.

Mitigation: Developers can curate and rebalance training datasets to reduce over/under-representation of specific groups, and apply automated bias-detection filters that flag or adjust generated outputs (e.g., diversifying results for prompts like "a doctor" or "a CEO") before they're shown to the user or allowed to be shared.
