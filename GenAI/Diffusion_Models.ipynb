# Diffusion Models (Stable Diffusion Intro)

Diffusion Models learn to gradually remove noise from random data, allowing them to generate highly realistic images, and Stable Diffusion uses this idea to create images directly from text prompts.

```python
# Diffusion Models are generative AI models that create new images by:

# Adding noise to images (Forward Diffusion)
# Learning how to remove noise (Reverse Diffusion)

# After training, the model can generate completely new images from random noise
```

```python
# Imagine you have a clear photo: cat_Image

# Add noise repeatedly:
# Original Image > Slight Noise > More Noise > Heavy Noise > Pure Random Noise

# Now train AI to reverse the process:
# Random Noise > Remove Noise > Remove More Noise > Remove More Noise > Clear Cat Image

# This is Diffusion
```

```python
# Forward Diffusion

# Forward diffusion means: Image > Noise
# Clear image > 10% noise > 30% noise > 70% noise > 100% noise
```

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()

image = x_train[0]

plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.show()
```

```python
# Add Noise

noise = np.random.normal(0, 50, image.shape)  # create random noise with Mean = 0, Std Dev = 50 
noisy_image = image + noise

noisy_image = np.clip(noisy_image, 0, 255) #clip values to valid pixel range

plt.imshow(noisy_image, cmap='gray')

plt.title("Noisy Image")
plt.show()
```

Reverse Diffusion


```python
# Means: Noise > Image

# Model learns:
# If image contains noise
# remove it gradually
```

```python
# Forward Process
# Image > Noise > More Noise > Pure Noise

# Reverse Process
# Pure Noise > Less Noise > Less Noise > Generated Image   
```

```python
# Why Diffusion Models Are Generative?

# GAN: Noise > Generator > Image
# Diffusion: Noise > Denoising > Image

# Both create new images
```

####  Stable Diffusion

```python
# Most famous diffusion model

# Used for: AI Art, Posters, Product Design, Illustrations, Concept Art
```

```python
# What is Latent Diffusion?
# Normal diffusion works directly on image pixels
# Very slow

# Stable Diffusion uses:
# Image > Compressed Representation > (Latent Space) > Diffusion > Image

# Similar to Autoencoder compression.
# Because of this: Faster, Less memory, Better quality
```

### Text-to-Image Generation

```python
# Input: "A cute puppy wearing sunglasses"

# Output: New image generated by AI.

# Process: Text Prompt, Text Encoder, Latent Space, Diffusion Process, Generated Image
```

```python
# Practical Using Stable Diffusion

# !pip install diffusers
# !pip install torch torchvision transformers
```

```python
# Generate Image From Text
from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5"
)

pipe = pipe.to("cuda")

image = pipe(
    "A cute puppy wearing sunglasses"
).images[0]

image.show()
```

Why Diffusion Models Replaced GANs?


```python
# GAN Problems

# Sometimes GAN produces: Strange faces, Missing eyes, Distorted images
# Training is difficult.
# Generator and Discriminator must remain balanced.
```

```python
# Diffusion Advantages

# Stable trainin, Better image quality, More diverse images, Easier scaling, Better text-to-image generation
```

```python
# GAN                       vs                      Diffusion
# Fast generation	                            Slower generation
# Difficult training                        	Stable training
# May collapse	                                More diverse outputs
# Good quality	                                Excellent quality
# Older approach	                            Modern approach
```
