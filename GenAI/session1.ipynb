Q2. Discriminative vs Generative Models
Aspect: Definition
Discriminative Model: Learns the boundary between classes; models P(y|x), predicting a label given an input.

Generative Model: Learns the underlying data distribution itself; models P(x) or P(x,y), allowing it to generate new data.

Aspect: Main Use
Discriminative Model: Classification and prediction tasks.
Generative Model: Creating new content such as text, images, or audio.

Aspect: Example Task
Discriminative Model: Classifying an email as spam or not spam.
Generative Model: Generating a new image of a face that doesn't belong to any real person.

Aspect: Example Model
Discriminative Model: Logistic Regression, SVM, standard CNN classifier.
Generative Model: GPT (text), GANs, VAEs, Stable Diffusion (images).
