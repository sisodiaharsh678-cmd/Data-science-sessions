```python
#simple_generator 

import numpy as np 

def simple_generator(random_vector):
    rng = np.random.default_rng(seed=int(np.sum(random_vector)*1000) % (2**32))
    fake_image = rng.random((28,28))
    return fake_image


#test it 

random_input = np.random.rand(100)
fake_digit = simple_generator(random_input)
print(fake_digit.shape)
print(fake_digit.min(), fake_digit.max())
```

```python
class Discriminator:
    def __init__(self, threshold=0.4):
        self.threshold = threshold  

    def discriminate(self, image):
    
        mean_pixel = np.mean(image)
        probability_real = max(0.0, min(1.0, 1 - (mean_pixel / self.threshold) * 0.5))
        return probability_real

# Test it
discriminator = Discriminator()
score = discriminator.discriminate(fake_digit)
print(f"Discriminator score for fake image: {score:.3f}")
```

```python
def simulate_one_gan_cycle():

    random_vector = np.random.rand(100)
    fake_image = simple_generator(random_vector)

  
    discriminator = Discriminator()
    score = discriminator.discriminate(fake_image)

    
    print(f"Fake image mean pixel value: {np.mean(fake_image):.3f}")
    print(f"Discriminator score (probability 'real'): {score:.3f}")

    if score < 0.5:
        print("Result: Discriminator correctly identified the image as FAKE.")
    else:
        print("Result: Discriminator was FOOLED — it thought the fake image was REAL.")

simulate_one_gan_cycle()
```

DCGAN vs StyleGAN vs CycleGAN — AI-generated explanation


DCGAN was the first architecture to make GAN training stable for images by using convolutional layers instead of plain dense layers, establishing it as the foundational blueprint most later image GANs build on. StyleGAN builds on that foundation but adds a style-based generator that injects learned 'style' information at multiple layers, letting it control coarse features (like pose) separately from fine details (like skin texture) — this is what makes its generated faces so sharp and controllable. CycleGAN solves a different problem entirely: translating images between two domains (e.g., photos↔paintings) without needing paired examples, using two generators and a cycle-consistency loss that enforces 'translate there and back should return the original image.
