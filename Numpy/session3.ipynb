```python
import numpy as np 

harsh = np.array([8000,10000,9000,8000])
raj = np.array([9000,8000,9800,10000])

print("Addition       :", harsh + raj)
print("Subtraction    :", harsh - raj)
print("Multiplication :", harsh * raj)
print("Division       :", (harsh / raj).round(2))
```

```python
import numpy as np

A = np.array([[8, 3, 1],   # User preferences
              [2, 7, 5],
              [4, 4, 9]])

B = np.array([[9, 6, 3],   # Song popularity
              [5, 8, 4],
              [2, 4, 9]])

print(np.dot(A, B))      
print(np.matmul(A, B))   
```

```python
import numpy as np 


image = np.array ([[120, 200,  85, 255],
                  [ 45, 178, 210,  90],
                  [230,  60, 145, 175],
                  [ 80, 195,  35, 220]])

print(image)  
print(image.T)
print(image.mean())
print(np.median(image))  
print(image.std().round(2))
print(image.var().round(2))          
```

```python
import numpy as np 

grid = np.array([[4.5, 3.2, 2.8],
                 [3.2, 4.1, 3.6],
                 [2.8, 3.6, 4.3]])

print(np.linalg.det(grid).round(2))  
print(np.linalg.inv(grid).round(2))   

eigenvalues, eigenvectors = np.linalg.eig(grid)
print(eigenvalues.round(4))
print(eigenvectors.round(4))
```
