```python
import cv2
import sys 
```

```python
image_path = "food.jpg"
```

```python
image = cv2.imread(image_path)
```

```python
if image is None:
    print("Error: Image not found. Please check the file path.")
    sys.exit()
```

```python
#resized 

resized = cv2.resize(image , (256,256) , interpolation=cv2.INTER_AREA)
```

```python
#split B ,G , R
B, G, R = cv2.split(resized)
```

```python
b_avg = round(B.mean(), 2)
g_avg = round(G.mean(), 2)
r_avg = round(R.mean(), 2)

print("Average Blue Intensity:", b_avg)
print("Average Green Intensity:", g_avg)
print("Average Red Intensity:", r_avg)
```

```python
cv2.imshow("Resized Image", resized)
cv2.imshow("Blue Channel", B)
cv2.imshow("Green Channel", G)
cv2.imshow("Red Channel", R)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
