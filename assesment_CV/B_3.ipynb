```python
import cv2
import sys

# Step 1: Load the image
image = cv2.imread("food.jpg")

if image is None:
    print("Error: Image not found. Please check the file path.")
    sys.exit()
```

```python
h , w = image.shape[:2]
```

```python
#rotate 90 degree 

center  = (w//2 , h//2)
angle = -90
scale = 1

M = cv2.getRotationMatrix2D(center, angle, scale)
rotated = cv2.warpAffine(image , M , (w , h))

cv2.imshow("Step 1 - Rotated", rotated)
```

```python
#crop lower-right 
rh, rw = rotated.shape[:2]
cropped = rotated[rh//2:rh, rw//2:rw]

cv2.imshow("Step 2 - Cropped (Lower-Right)", cropped)
```

```python
#flip 
flipped = cv2.flip(cropped, 1)

cv2.imshow("Step 3 - Flipped", flipped)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#gaussian blur before edge detection 

blurred = cv2.GaussianBlur(flipped, (5, 5), sigmaX=0)
```

```python
#canny edge detection 
edges = cv2.Canny(blurred, 50, 150)
```

```python
cv2.imshow("Label Edges", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#save 
cv2.imwrite("label_edges.jpg", edges)
print("Saved as label_edges.jpg")
```
