```python
def compute_gredient(f , x , h=1e-5):
    return (f(x+h) - f(x-h) / (2 * h))

f = lambda x:x**2
print(compute_gredient(f, 3))  
```

```python
def compute_partial(f, point, var_index, h=1e-5):
    point_plus = list(point)
    point_minus = list(point)
    point_plus[var_index] += h
    point_minus[var_index] -= h
    return (f(*point_plus) - f(*point_minus)) / (2 * h)

f = lambda x, y: x**2 + y**2

df_dx = compute_partial(f, (3, 4), var_index=0)   
df_dy = compute_partial(f, (3, 4), var_index=1)   

print("df/dx at (3,4):", df_dx)   
print("df/dy at (3,4):", df_dy)   
```

```python
def gradient(w, h=1e-5):
    f = lambda w: (w - 5)**2
    return (f(w + h) - f(w - h)) / (2 * h)

w = 0
learning_rate = 0.1
w = w - learning_rate * gradient(w)
print("Updated w:", w) 
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**4 - 3*x**3 + 2

x = np.linspace(-2, 4, 400)
y = f(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('f(x) = x^4 - 3x^3 + 2')
plt.grid(True)
plt.show()
```
