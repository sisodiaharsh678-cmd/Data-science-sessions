```python
def neuron_output(input_value , weight , bias):
    return input_value * weight + bias 
```

```python
likes = [50, 120, 80, 200, 30]
weight = 0.8
bias = 2

for like in likes:
    output = neuron_output(like, weight, bias)
    print(f"Likes: {like} -> Output: {output}")
```

```python
import numpy as np

n = 10

zeros = [0] * n
random_vals = np.random.uniform(-1, 1, n)
xavier_vals = np.random.normal(0, 1/np.sqrt(n), n)

print("Zeros:", zeros)
print("Random (-1 to 1):", random_vals)
print("Xavier init:", xavier_vals)
```

```python
output1 = neuron_output(0, 0.5, 0)   
output2 = neuron_output(0, 0.5, 5)   

print("Output with bias=0:", output1)
print("Output with bias=5:", output2)
```
