```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

X = np.array([1.0, 2.0, 3.0])

W1 = np.array([[0.2, -0.1], [0.4, 0.3], [-0.5, 0.2]])  # shape (3,2)
W2 = np.array([0.6, -0.3])  # shape (2,)

hidden = relu(np.dot(X, W1))
output = sigmoid(np.dot(hidden, W2))

print("Hidden activations:", hidden)
print("Output:", output)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

X = np.array([1.0, 2.0, 3.0])
W1 = np.array([[0.2, -0.1], [0.4, 0.3], [-0.5, 0.2]])
W2 = np.array([0.6, -0.3])

hidden = relu(np.dot(X, W1))
output = sigmoid(np.dot(hidden, W2))
output = np.array([output])  #

fig, axs = plt.subplots(1, 3, figsize=(10, 3))
axs[0].bar(range(len(X)), X)
axs[0].set_title('Input')
axs[1].bar(range(len(hidden)), hidden)
axs[1].set_title('Hidden Layer')
axs[2].bar(range(len(output)), output)
axs[2].set_title('Output')
plt.tight_layout()
plt.show()
```

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)


X = np.array([
    [1.0, 2.0, 3.0],
    [0.5, -1.0, 2.0],
    [2.0, 0.0, -1.0]
])

W1 = np.random.uniform(-1, 1, (3, 4))  
W2 = np.random.uniform(-1, 1, (4, 1))  

hidden = relu(np.dot(X, W1))
output = sigmoid(np.dot(hidden, W2))

for i, out in enumerate(output):
    print(f"Input {i+1}: {X[i]} - Output: {out}")
```
