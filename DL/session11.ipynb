```python
import tensorflow as tf 
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense

(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
X_train, X_test = X_train/255.0, X_test/255.0

model_sgd = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128 , activation='relu'),
    Dense(10,activation='softmax')
])

model_sgd.compile(optimizer='sgd', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

history_sgd = model_sgd.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=5, batch_size=32)

print("SGD - Final val_accuracy:", round(history_sgd.history['val_accuracy'][-1], 4))
print("SGD - Final val_loss:", round(history_sgd.history['val_loss'][-1], 4))
```

```python
model_rmsprop = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model_rmsprop.compile(optimizer='rmsprop', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

history_rmsprop = model_rmsprop.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=5, batch_size=32)

print("RMSprop - Final val_accuracy:", round(history_rmsprop.history['val_accuracy'][-1], 4))
print("SGD was:", round(history_sgd.history['val_accuracy'][-1], 4))
```

```python
from tensorflow.keras.optimizers import Adam

model_adam = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model_adam.compile(optimizer=Adam(learning_rate=0.0005),
                    loss='sparse_categorical_crossentropy', metrics=['accuracy'])

history_adam = model_adam.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=5, batch_size=32)

print("Adam (lr=0.0005) - Final val_accuracy:", round(history_adam.history['val_accuracy'][-1], 4))
```
