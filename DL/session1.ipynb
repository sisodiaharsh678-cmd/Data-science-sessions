Low LR (0.001–0.003): smooth but painfully slow. Loss decreases steadily but takes hundreds of epochs to separate the classes.

Task 3 — 3 Apps Using Deep Learning
1. Instagram — Feed Recommendation

Instagram uses deep learning to analyze your likes, saves, and watch time to decide what posts to show next. A deep neural network learns your behavior patterns and predicts which content will keep you engaged longest.
2. Spotify — Song Recommendation

Spotify uses deep learning to analyze audio features (tempo, pitch, energy) of songs you like and finds similar ones. It also uses collaborative filtering — if 1000 users with similar taste liked a song, it recommends it to you.
3. Zomato — Food Recommendation

Zomato uses deep learning to analyze your past orders, time of day, location, and ratings to suggest restaurants. It also uses image recognition deep learning to verify food photos uploaded by restaurants.

Task 4 — Deep Learning Workflow for Movie Recommendation
Stage 1 — Data Collection

Collect user watch history, ratings, and movie metadata (genre, cast, director).
Stage 2 — Preprocessing

Clean data, encode user and movie IDs, normalize ratings, split 80/20 train/test.
Stage 3 — Model Building

Build Neural Collaborative Filtering — user + movie embeddings → hidden layers → predicted rating output.
Stage 4 — Training

Use MSE loss, Adam optimizer, train 50–100 epochs with early stopping.
Stage 5 — Evaluation

Measure RMSE and Precision@K to check recommendation quality.
Stage 6 — Deployment

Save model → Flask API → integrate into app → User ID in, Top 10 movies out → retrain monthly.
