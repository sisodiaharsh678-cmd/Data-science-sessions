```python
def calculate_mse(predictions, targets):
    n = len(predictions)
    squared_errors = [(p - t) ** 2 for p, t in zip(predictions, targets)]
    return sum(squared_errors) / n

preds = [3, 5, 2.5, 7]
targets = [2.5, 5, 4, 8]
print("MSE:", calculate_mse(preds, targets))
```

```python
actual_ratings = [4, 3, 5, 2, 1]
predicted_ratings = [5, 2, 4, 2, 1]

errors = [abs(a - p) for a, p in zip(actual_ratings, predicted_ratings)]
mae = sum(errors) / len(errors)
print("MAE:", mae)
```

```python
import numpy as np

def binary_cross_entropy(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    loss = -(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
    return np.mean(loss)

y_true = [1, 0, 1]
y_pred = [0.9, 0.2, 0.7]
print("BCE Loss:", binary_cross_entropy(y_true, y_pred))
```

```python
import numpy as np

y_true = [0, 0, 1]
y_pred = [0.2, 0.3, 0.5]

loss = -np.sum([t * np.log(p) for t, p in zip(y_true, y_pred)])
print("Categorical Cross-Entropy Loss:", loss)
```

Scenario: Predicting whether a restaurant photo shows "food" vs "not food"
(for content moderation before it's posted) is a binary classification problem.

Loss function: Binary Cross-Entropy
Why: The output is a single probability (0-1) for two possible classes,
and BCE penalizes confident wrong predictions heavily -- important for
moderation accuracy, where a confidently wrong prediction (e.g., 0.95
probability "food" for a non-food image) should be corrected fast.
