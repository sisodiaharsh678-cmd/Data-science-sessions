```python
import numpy as np
import matplotlib.pyplot as plt

def parametric_relu(x, a):
    return np.where(x > 0, x, a * x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

x = np.linspace(-5, 5, 200)

plt.plot(x, relu(x), label='ReLU')
plt.plot(x, leaky_relu(x), label='Leaky ReLU (0.01)')
plt.plot(x, parametric_relu(x, 0.2), label='PReLU (a=0.2)')
plt.legend()
plt.title('ReLU vs Leaky ReLU vs PReLU')
plt.grid(True)
plt.show()

    
```

```python
import numpy as np

def softmax(scores):
    exp_scores = np.exp(scores)
    return exp_scores / np.sum(exp_scores)

scores = [2.0, 1.0, 0.1, 0.5]
probs = softmax(scores)
print("Probabilities:", np.round(probs, 3))
```

```python
def hard_sigmoid(x):
    return max(0, min(1, 0.2 * x + 0.5))

for x in [-3, -1, 0, 1, 3]:
    print(f"x={x} -> {hard_sigmoid(x)}")
```

(a) Hidden dense layer -> ReLU
    Fast to compute and avoids vanishing gradients, so deep hidden layers train efficiently.

(b) Output layer, 5 classes -> Softmax
    Converts raw scores into a probability distribution across all 5 classes that sums to 1.

(c) Output layer, binary classification -> Sigmoid
    Naturally outputs a single probability between 0 and 1 for a two-class decision.
