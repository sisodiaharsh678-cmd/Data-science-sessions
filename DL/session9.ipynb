```python
import tensorflow as tf 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([
    Dense(4, activation = 'relu' , input_shape=(3,)),
    Dense(1, activation = 'sigmoid')
])

model.summary()
```

```python
import numpy as np

model = Sequential([
    Dense(4, activation = 'relu' , input_shape=(3,)),
    Dense(1, activation = 'sigmoid')
])

model.summary()

model.compile(optimizer='sgd' , loss='mse')

X_dummy = np.array([[1.0 , 2.0 , 3.0]])
y_dummy = np.array([[1.0]])

print("Weights BEFORE training:")
print(model.layers[0].get_weights()[0])

model.fit(X_dummy, y_dummy, epochs=1, verbose=0)

print("\nWeights AFTER training:")
print(model.layers[0].get_weights()[0])
```

```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

value = 0.9
for layer in range(1, 11):
    value = sigmoid(value)
    print(f"After layer {layer}: {value:.6f}")
```

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)


np.random.seed(1)
value = np.array([1.0])

for layer in range(1, 11):
    large_weight = np.random.uniform(5, 10)   
    value = relu(value * large_weight)
    print(f"After layer {layer}: {value}")
```
