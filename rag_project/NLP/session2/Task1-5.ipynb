```python
#task 2 

import string 

def remove_punctuation(review):
    return review.translate(str.maketrans('' , '' , string.punctuation))

review = "Wow! This phone is amazing. Battery life? Excellent! Worth every rupee, seriously."
print(remove_punctuation(review))    
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
#task -4 

def remove_stopwords(message):
    stop_words = set(stopwords.words('english'))

    words = message.lower().split()

    flitered = [word for word in words if word not in stop_words]

    return flitered 

message = "Hey are you coming to the party tonight at the new restaurant near the mall?"
print("Original:", message)
print("Cleaned:", remove_stopwords(message))
```

```python
#step -5 

import nltk
nltk.download('punkt')
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

hashtags = ['#playing', '#dancing', '#eating', '#running', '#coding', '#loving']

for tag in hashtags:
    word = tag.replace("#" , "")
    stemmed = stemmer.stem(word)
    print(f"{tag:<15} {stemmed}")
```
