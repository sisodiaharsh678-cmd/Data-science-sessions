```python
import nltk
nltk.download('stopwords')
```

```python
#stopswords
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

words = ['natural' , 'language' , 'is','a','of','the']
new_words = [i for i in words if i not in stop_words]
new_words
```

```python
#stemming - convert into base word 
#portersteaming 
from nltk.stem import PorterStemmer

steaming = PorterStemmer()
words = ['running', 'happily','better']

for i in words:
    print(i,"--" , steaming.stem(i))
```

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

steaming = WordNetLemmatizer()
words = ['running', 'happily', 'better']

for i in words:
    print(i, "--", steaming.lemmatize(i))
```
