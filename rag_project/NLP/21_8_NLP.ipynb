```python
# stopwords
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))
words = ["natural", "language", "is", "a", "field", "of", "the", "science"]

new_words = [i for i in words if i not in stop_words]
new_words
```

```python
# stemming - convert into base word
# portersteaming
from nltk.stem import PorterStemmer

steaming = PorterStemmer()
words = ["running", "studies", "happily", "better", "flies"]

for i in words:
    print(i, "--", steaming.stem(i))
```

```python
# stemming - convert into base word
# portersteaming
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
words = ["running", "studies", "happily", "better", "flies", "enjoying"]

for i in words:
    print(i, "--", lemmatizer.lemmatize(i))
```

```python
lemmatizer.lemmatize("best", pos="v")
lemmatizer.lemmatize("freezing", pos="v")
```

```python
sample_text = """
Natural language processing (NLP) is a field of artificial intelligence that focuses on the interaction
between computers and humans through natural language. The ultimate goal of NLP is to read, decipher,
understand, and make sense of human language in a way that is both valuable and meaningful.
"""
```

```python
sentence = sent_tokenize(sample_text)
sentence
```

```python
words = [word_tokenize(i) for i in sentence]
words
```

```python
cleaned_words = [
    [re.sub(r"[^a-zA-Z0-9]", "", word.lower()) for word in sentence]
    for sentence in words
]
cleaned_words
```

```python
stop_words = set(stopwords.words("english"))

new_words = [
    [word for word in sentence if word not in stop_words] for sentence in cleaned_words
]
new_words
```

```python
stemmer = PorterStemmer()
stemmed_words = [[stemmer.stem(word) for word in sentence] for sentence in new_words]
```

```python
stemmed_words
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatizer_words = [[lemmatizer.lemmatize(word) for word in sentence] for sentence in new_words]

```

```python
lemmatizer_words
```

```python
for sentence in stemmed_words:
    print(" ".join(sentence))
```
