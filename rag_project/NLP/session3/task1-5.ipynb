```python
#Task 1 — Vocabulary from Instagram captions

import string 

captions = [
    "Sunset vibes at the beach!",
    "Best day ever with my best friends!",
    "Coffee and good vibes only"
]

def build_vocabulary(captions):
    vocab = set()
    for caption in captions:
        cleaned = caption.translate(str.maketrans('' ,'',string.punctuation))
        words = cleaned.lower().split()
        vocab.update(words)
    return sorted(vocab)

vocabulary = build_vocabulary(captions)
print("Vocabulary:", vocabulary)
```

```python
#task -2 Bag of Words matrix using CountVectorizer

from sklearn.feature_extraction.text import CountVectorizer

reviews = ['Loved the pizza', 'Service was slow', 'Great ambience']

vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(reviews)

print("Vocab : " , vectorizer.get_feature_names_out())
print("BoW Matrix:\n", bow_matrix.toarray())
```

```python
#Task 3 — Word frequency across WhatsApp messages (with stopword removal)

import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords 
from collections import Counter 

messages = [
    "Hey are you coming to the party",
    "The party is at my place tonight",
    "Are you bringing snacks to the party"
]

stop_words = set(stopwords.words('english'))

def word_frequency(messages , stop_words):
    freq={}
    for message in messages:
        words = message.lower().split()
        for word in words:
            if word not in stop_words:
                freq[word]= freq.get(word , 0) +1
    return freq     

result = word_frequency(messages, stop_words)
print("Word Frequency:", result)           
```

```python
#Task 4 — TF-IDF vectors using TfidfVectorizer

from sklearn.feature_extraction.text import TfidfVectorizer

reviews = [
    "The quality of this product is excellent",
    "Poor quality, broke within a week",
    "Amazing quality and great packaging"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(reviews)

feature_names = vectorizer.get_feature_names_out()
quality_index = list(feature_names).index('amazing')

print("Feature names:", feature_names)

for i, review in enumerate(reviews):
    tfidf_score = tfidf_matrix[i, quality_index]
    print(f"Review {i+1}: '{review}' -  TF-IDF('quality') = {tfidf_score:.4f}")
```
