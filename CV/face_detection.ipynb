# Display Webcam (Hello Camera)

```python
import cv2
camera = cv2.VideoCapture(0)    #Open webcam

while True:
    success, frame = camera.read()  #Current Frame
    if not success:
        break
    cv2.imshow("Webcam", frame) #Displays image
    if cv2.waitKey(1) & 0xFF == ord('q'):   #Wait 1 millisecond
        break
    
camera.release()
cv2.destroyAllWindows()
```

# Convert Webcam into Black & White

```python
import cv2
camera = cv2.VideoCapture(0)

while True:
    ret, frame = camera.read()
    if not ret:
        break
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    cv2.imshow("Gray Camera", gray)
    if cv2.waitKey(1)==ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```

# Mirror Camera

```python
import cv2
camera=cv2.VideoCapture(0)

while True:
    ret,frame=camera.read()
    if not ret:
        break
# 1-Horizontal flip, 0-Vertical, -1-Both
    mirror=cv2.flip(frame,1)
    cv2.imshow("Mirror",mirror)
    if cv2.waitKey(1)==ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```

# Draw Shapes on Live Camera

```python
import cv2
camera=cv2.VideoCapture(0)

while True:
    ret,frame=camera.read()
    cv2.rectangle(frame,(50,50),(250,200),(0,255,0),3)
    cv2.circle(frame,(400,200),60,(255,0,0),3)
    cv2.line(frame,(0,0),(640,480),(0,0,255),2)
    cv2.putText(frame,
                "TOPS Technologies",
                (50,450),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (255,255,0),
                2)
    cv2.imshow("Drawing",frame)

    if cv2.waitKey(1)==ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```

# Face Detection

```python
import cv2

face_detector=cv2.CascadeClassifier(
cv2.data.haarcascades+
'haarcascade_frontalface_default.xml')
camera=cv2.VideoCapture(0)

while True:
    ret,frame=camera.read()
    gray=cv2.cvtColor(frame,
                      cv2.COLOR_BGR2GRAY)
    faces=face_detector.detectMultiScale(
        gray,
        1.1,
        5
    )

    for x,y,w,h in faces:
        cv2.rectangle(frame,
                      (x,y),
                      (x+w,y+h),
                      (0,255,0),
                      2)
    cv2.imshow("Face Detection",frame)

    if cv2.waitKey(1)==ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```

# Face Counter

```python
count=len(faces)

cv2.putText(frame,
            f"Faces: {count}",
            (20,40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0,255,0),
            2)
```

# Motion Detection (Frame Difference)

```python
import cv2
camera=cv2.VideoCapture(0)
ret,previous=camera.read()

previous=cv2.cvtColor(previous,
                      cv2.COLOR_BGR2GRAY)

while True:
    ret,current=camera.read()
    gray=cv2.cvtColor(current,
                      cv2.COLOR_BGR2GRAY)
    difference=cv2.absdiff(previous,
                           gray)
    cv2.imshow("Motion",
               difference)
    previous=gray

    if cv2.waitKey(1)==ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```

# Edge Detection (Canny)

```python
import cv2
camera = cv2.VideoCapture(0)
while True:
    ret, frame = camera.read()
    if not ret:
        break
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 100, 200)

    cv2.imshow("Original", frame)
    cv2.imshow("Edges", edges)
    if cv2.waitKey(1) == ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```

# Save Image When Pressing 'S'

```python
import cv2
camera = cv2.VideoCapture(0)
count = 1

while True:
    ret, frame = camera.read()
    if not ret:
        break
    cv2.imshow("Camera", frame)
    key = cv2.waitKey(1)

    if key == ord('s'):
        filename = f"photo_{count}.jpg"
        cv2.imwrite(filename, frame)
        print("Saved:", filename)
        count += 1

    elif key == ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```

# Blur Faces (Privacy Protection)

```python
import cv2

face_detector = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)
camera = cv2.VideoCapture(0)

while True:
    ret, frame = camera.read()
    if not ret:
        break
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face_detector.detectMultiScale(gray, 1.1, 5)

    for x, y, w, h in faces:
        face = frame[y:y+h, x:x+w]
        blurred = cv2.GaussianBlur(face, (51, 51), 30)
        frame[y:y+h, x:x+w] = blurred
    cv2.imshow("Blur Faces", frame)

    if cv2.waitKey(1) == ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```
