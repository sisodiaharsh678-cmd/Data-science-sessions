```python
resize = cv2.resize(image , (500,200))
cv2.imshow("this is image" , image) 
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
#convert to gray scale 
gray_scale = cv2.cvtColor(image ,cv2.COLOR_RGB2BGR)
cv2.imshow("this is image" , gray_scale) 
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
b , g , r = cv2.split(image)
cv2.imshow("blue" , b) 
cv2.imshow("green" , g) 
cv2.imshow("red" , r) 
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
import matplotlib.pyplot as plt

b, g, r = cv2.split(image)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].imshow(b, cmap='gray')
axes[0].set_title("Blue Channel")
axes[0].axis('off')

axes[1].imshow(g, cmap='gray')
axes[1].set_title("Green Channel")
axes[1].axis('off')

axes[2].imshow(r, cmap='gray')
axes[2].set_title("Red Channel")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
#merged image again from b , g , r
merged = cv2.merge([b,g,r])
cv2.imshow("merged" , merged) 
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
#image transfromation 
```

```python
h , w = image.shape[:2]
cv2.imshow("merged" , w) 
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
center = (w//2. ,h//2)
scale = 1 #1-same-size , 2=zoom in 
angle = 45

M= cv2.getRotationMatrix2D(center , angle , scale)
rotated = cv2.warpAffine(image, M, (w,h))

cv2.imshow("merged" , rotated) 
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
#crop image
crop = image[50:300 , 100:450]
cv2.imshow("merged" , crop) 
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
#flip image 

flip_img = cv2.flip(image,-1)
cv2.imshow("merged" , flip_img) 
cv2.waitKey(0)
cv2.destroyAllWindows() 

# 1- horizotal 
# -1 = vertical 
# 0 - both axex
```
