```python
import cv2
img = cv2.imread(r"/Users/harshsisodiya678/TOPS/Sessions/CV/Screenshot 2026-07-23 at 3.28.59 PM.png")
cv2.imshow('orignal',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
blured = cv2.blur(img , (7,7))
cv2.imshow('orignal',blured)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gua_blured = cv2.GaussianBlur(img , (7,7),0)
cv2.imshow('orignal',gua_blured)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
med_blured = cv2.medianBlur(img ,7)
cv2.imshow('orignal',med_blured)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
