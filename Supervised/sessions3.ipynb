```python
import pandas as pd
import numpy as np

df = pd.read_csv('ipl_scores_missing.csv')
print(df.isnull().sum()) 

df.info()
```

```python
num_cols =['match_id' , 'overs' , 'wickets' , 'runs_scored' , 'run_rate','target']

df[num_cols]=df[num_cols].fillna(df[num_cols].median())

cat_cols = ['batting_team' , 'bowling_team' , 'venue' , 'result']

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum()
```

```python
df.head()
```

```python
venue_dummies = pd.get_dummies(df['venue'], drop_first=True)
df = pd.concat([df, venue_dummies], axis=1)
df = df.drop(columns=['venue'])

print(df.shape)
print(df.head(2))
```

```python
df.head(2)
```

```python
from sklearn.preprocessing import LabelEncoder 

le = LabelEncoder()
cat_cols = ['batting_team' , 'bowling_team' , 'result']
for col in cat_cols:
    df[col]=le.fit_transform(df[col])

df.head()    
```
