```python
from sklearn.datasets import load_iris 
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

iris = load_iris()
X , y = iris.data , iris.target

model = KNeighborsClassifier(n_neighbors = 5)
scores = cross_val_score(model , X , y , cv=5)

print("Fold scores:", scores)
print("Average Accuracy:", scores.mean())
```

```python
from sklearn.datasets import load_wine 
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score

wine = load_wine()
X , y = wine.data , wine.target

model = DecisionTreeClassifier(random_state=42)
scores = cross_val_score(model, X, y, cv=10)

for i , score in enumerate(scores , 1):
    print(f"Fold {i}: {score:.4f}")

print(f"\nAverage Accuracy: {scores.mean():.4f}")   
```

```python
from sklearn.datasets import load_wine
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

wine = load_wine()
X, y = wine.data, wine.target

model = RandomForestClassifier(n_estimators=100, random_state=42)

for cv_val in [3,5,10]:
    scores = cross_val_score(model , X , y , cv=cv_val)
    print(f"cv={cv_val} → Mean={scores.mean():.4f}, Std={scores.std():.4f}")
```

```python
from sklearn.model_selection import cross_val_score 
import numpy as np 

def evaluate_with_cv(model, X, y, cv=5, scoring='accuracy'):

    scores = cross_val_score(model, X, y, cv=cv, scoring=scoring)
    return {
        'scores': scores,
        'mean'  : scores.mean(),
        'std'   : scores.std()
 
    }

from sklearn.datasets import load_iris
from sklearn.svm import SVC

iris = load_iris()
X, y = iris.data, iris.target

result = evaluate_with_cv(SVC(), X, y, cv=5)
print("Mean Accuracy:", result['mean'])
print("Std Deviation:", result['std'])    
    
    
```
