```python
import pandas as pd 
import numpy as np 
from sklearn.preprocessing import StandardScaler 

np.random.seed(42)
n=100

# Create IPL dataset
ipl = pd.DataFrame({
    'player_name'  : ['Player_' + str(i) for i in range(n)],
    'matches'      : np.random.randint(5, 100, n),
    'runs'         : np.random.randint(0, 3000, n),
    'wickets'      : np.random.randint(0, 100, n),
    'strike_rate'  : np.round(np.random.uniform(60, 200, n), 2),
    'average'      : np.round(np.random.uniform(10, 60, n), 2),
    'economy_rate' : np.round(np.random.uniform(5, 12, n), 2),
})

ipl.head(5)

numeric_cols = ['matches', 'runs', 'wickets', 'strike_rate', 'average', 'economy_rate']

ipl.info()
ipl.isnull().sum()

scaler = StandardScaler()
scaled_values = scaler.fit_transform(ipl[numeric_cols])

scaled_values 

```

```python
from sklearn.preprocessing import MinMaxScaler
np.random.seed(42)
n = 100

flipkart = pd.DataFrame({
    'product'    : ['Product_' + str(i) for i in range(n)],
    'price'      : np.random.randint(200, 50000, n),
    'rating'     : np.round(np.random.uniform(1.0, 5.0, n), 1),
    'num_reviews': np.random.randint(10, 10000, n),
    'discount'   : np.random.randint(0, 80, n),
})

numeric_cols = ['price', 'rating', 'num_reviews', 'discount']

flipkart.head()

print("before scalling ")

flipkart[numeric_cols].min() , flipkart[numeric_cols].max()

scaler = MinMaxScaler()
scaled_values = scaler.fit_transform(flipkart[numeric_cols])

flipkart_scaled = flipkart.copy()
flipkart_scaled[numeric_cols] = scaled_values

scaled_values
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.feature_selection import RFE

df = pd.read_csv("zomato_restaurants.csv")

# Direct Drop - sab string/text columns
X = df.drop(columns=['restaurant_id', 'restaurant_name', 'city', 
                      'cuisines', 'average_cost_for_two'])
y = df['average_cost_for_two']

print(X.columns.tolist())  # verify karo sirf numeric hai
print(X.shape)

# RFE
model = DecisionTreeRegressor(random_state=42)
rfe = RFE(estimator=model, n_features_to_select=3)
rfe.fit(X, y)

selected_features = X.columns[rfe.support_].tolist()
print("Selected Features:", selected_features)
```

```python
import pandas as pd

df = pd.read_csv('bookmyshow_ratings.csv')

Q1  = df['user_rating'].quantile(0.25)
Q3  = df['user_rating'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_clean = df[(df['user_rating'] >= lower) & (df['user_rating'] <= upper)]

print(f"Rows dropped: {len(df) - len(df_clean)}")




```
