Task-1 - 3 Apps with ML Feature


Zomato:
Zomato uses ML in its food recommendation system. Based on your past orders, time of day, and location, it recommends restaurants and dishes you are likely to order. This is an example of Supervised Learning because the model is trained on labeled historical data — what users ordered and rated.
Instagram:
Instagram uses ML in its Reels recommendation feature. It tracks what videos you watch fully, skip, like, or share and learns your preferences over time. This is an example of Reinforcement Learning because the app gets rewarded when you engage more and keeps optimizing to maximize your time on the app.
Flipkart:
Flipkart uses ML in its "Customers also bought" recommendation feature. It groups users with similar buying patterns and suggests products accordingly. This is an example of Unsupervised Learning because no one manually labeled which products go together — the model finds patterns on its own.

Task 2 — Classify Each Scenario

1. Netflix recommending movies based on watch history
→ Supervised Learning — the model learns from labeled historical data where movies you watched and rated act as training examples with clear feedback.
2. Spotify grouping similar songs into playlists
→ Unsupervised Learning — no one tells Spotify which songs belong together, it finds patterns in audio features like tempo, genre, and mood on its own without any labels.
3. A self-driving car learning to park by trial and error
→ Reinforcement Learning — the car learns by trying different actions, receiving a reward when it parks correctly and a penalty when it makes mistakes, improving over time through experience.

Task 4 — Spam Filter: Traditional vs ML

Traditional Programming approach:
In traditional programming, a developer manually writes fixed rules to detect spam. For example — "if the message contains words like FREE, WINNER, CLAIM, or PRIZE then mark it as spam." This approach works initially but has serious limitations. Every time spammers change their language or use new tricks, a programmer has to manually update the rules. It becomes impossible to maintain because there are infinite variations of spam messages and no set of hand-written rules can cover all of them.
Machine Learning approach:
In the machine learning approach, instead of writing rules manually, you collect thousands of examples of spam and non-spam WhatsApp messages and label them. The model learns on its own which patterns indicate spam — unusual word combinations, message length, sender behavior, frequency of messages, and more. When spammers change their tactics, you simply retrain the model with new labeled examples and it adapts automatically. The system gets smarter over time without anyone writing a single new rule.
