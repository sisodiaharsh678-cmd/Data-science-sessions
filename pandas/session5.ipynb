```python
import pandas as pd

today_orders = pd.DataFrame({
    'order_id': [201, 202, 203],
    'item'    : ['Burger', 'Pizza', 'Pasta'],
    'price'   : [199, 399, 349]
})

yesterday_orders = pd.DataFrame({
    'order_id': [101, 102, 103],
    'item'    : ['Biryani', 'Noodles', 'Sandwich'],
    'price'   : [249, 179, 129]
})

combined = pd.concat([today_orders, yesterday_orders], ignore_index=True)
print(combined)
```
