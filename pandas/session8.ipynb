```python
import pandas as pd 

df = pd.read_csv('zomato_ratings.csv')

df

Q1 = df['rating'].quantile(0.25)
Q3 = df['rating'].quantile(0.75)

IQR = Q3-Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['rating'] < lower) | (df['rating'] > upper)]
print("Outlier indices:", outliers.index.tolist())
print(outliers[['restaurant', 'rating']])
```

```python
import pandas as pd

df = pd.read_csv('paytm_transactions.csv')

lower = df['transaction_amount'].quantile(0.05)
upper = df['transaction_amount'].quantile(0.95)

df['transaction_amount'] = df['transaction_amount'].clip(lower, upper)
print(df['transaction_amount'].describe().round(2))
```
