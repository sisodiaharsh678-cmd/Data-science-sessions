Task 1 — Training vs Deployment 

Training is the phase where a machine learning model learns patterns from historical data — it is experimental, involves trying different algorithms, and can fail and retry freely. Deployment is when the trained model is finalized and made available to real users to serve predictions on new incoming data. For example, in Zomato, the recommendation model is first trained offline on millions of past orders, user preferences, and ratings. Once ready, it is deployed so that every time I open the app, it instantly shows me personalized food suggestions based on my history — without retraining every time I visit.

Task 3 — Online vs Offline for BookMyShow 

For a BookMyShow movie recommendation feature, I would use a hybrid approach — primarily offline (batch) deployment with some online elements. The model would be retrained periodically (daily or weekly) on new booking data, ratings, and trending movies, and recommendations would be pre-computed and stored for each user. This works well because movie preferences don't change every minute — a user's taste stays fairly stable. However, for real-time trending movies or newly released films, a lightweight online model would update recommendations instantly so users always see the latest releases.

```python
#Task 4 — Batch Prediction Function 

def batch_predict(review):
    print("batch predication results : ")
    for i , review in enumerate(review , 1):
        if "good" in review.lower():
            label = "positive"
        else:
            label = 'negative'
        print(f'review {i} : {label}')
        print(f" - {review}")  


reviews = [
    "The food was really good today",
    "Service was terrible and slow",
    "Good ambience but food was cold",
    "Worst experience ever never coming back",
    "The pizza was good and fresh",
    "Very disappointing meal today",
    "Staff was rude and unhelpful",
    "Good value for money loved it",
    "Food arrived late and was stale",
    "Everything was good highly recommend"
]

batch_predict(reviews)              
```
