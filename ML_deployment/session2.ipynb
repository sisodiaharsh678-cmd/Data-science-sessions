```python
#Task 1 — Train DecisionTreeClassifier + save with Pickle

import pickle 
from sklearn.tree import DecisionTreeClassifier

X = [
    [120, 0.8, 0.7],   # Pop
    [90, 0.4, 0.3],    # Classical
    [140, 0.9, 0.8],   # EDM
    [80, 0.3, 0.2],    # Classical
    [128, 0.85, 0.75], # Pop
    [150, 0.95, 0.85], # EDM
    [95, 0.35, 0.25],  # Classical
    [125, 0.82, 0.72], # Pop
]

y = ["Pop", "Classical", "EDM", "Classical", "Pop", "EDM", "Classical", "Pop"]

model = DecisionTreeClassifier()
model.fit(X , y)

with open("song_genre_model.pkl", "wb") as f:
    pickle.dump(model , f)  
```

```python
import pickle 

with open("song_genre_model.pkl", "rb") as f:
    loaded_model = pickle.load(f)

new_song = [[135, 0.88, 0.78]]
prediction = loaded_model.predict(new_song)

print(f"Predicted genre: {prediction[0]}")  
```

```python
import joblib 
from sklearn.neighbors import KNeighborsClassifier

X = [
    [15000, 180],   # Electronics (phone)
    [1200, 400],    # Fashion (jacket)
    [45000, 1500],  # Electronics (laptop)
    [800, 250],     # Fashion (shirt)
    [3000, 300],    # Electronics (earbuds)
    [1500, 500],    # Fashion (shoes)
]
y = ["electronics", "fashion", "electronics", "fashion", "electronics", "fashion"]


model = KNeighborsClassifier(n_neighbors=3)
model.fit(X , y)

joblib.dump(model, "product_category_model.joblib")

#load nd test
loaded_model = joblib.load("product_category_model.joblib")
new_product = [[20000, 200]]  # price=20000, weight=200g
prediction = loaded_model.predict(new_product)
print(f"Predicted category: {prediction[0]}")
```
