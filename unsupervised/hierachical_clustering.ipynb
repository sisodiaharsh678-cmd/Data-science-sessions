```python
#task- 1

import numpy as np 
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage


orders = np.array([
   [10, 2], [20, 5], [15, 3],
    [40, 8], [35, 7], [25, 6], 
])

n = len(orders)
distance_matrix = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        distance_matrix[i][j] = np.sqrt(
            (orders[j][0] - orders[i][0])**2 +
            (orders[j][1] - orders[i][1])**2
        )

print("Distance Matrix:")
print(np.round(distance_matrix, 2)) 
```

```python
#task-2

songs = np.array([
    [60, 30], [65, 25], [90, 55], [95, 60],
    [140, 85], [145, 90], [170, 95], [175, 92],
])

song_names = ['Slow Ballad', 'Soft Pop', 'Pop', 'Dance Pop',
              'EDM', 'Techno', 'Hard Rock', 'Metal']

linked  = linkage(songs, method = 'ward')

plt.figure(figsize=(8, 4))
dendrogram(linked, labels=song_names, leaf_rotation=45)
plt.title('Dendrogram — Ward Linkage')
plt.xlabel('Songs')
plt.ylabel('Distance')
plt.tight_layout()
plt.show()
```

```python
# task-3

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, method in zip(axes, ['single', 'complete', 'average']):
    linked = linkage(songs, method=method)
    dendrogram(linked, labels=song_names, leaf_rotation=45, ax=ax)
    ax.set_title(f'{method.capitalize()} Linkage')
    ax.set_xlabel('Songs')
    ax.set_ylabel('Distance')

plt.tight_layout()
plt.show()
```

#task-5

Best Linkage → Ward

Why?
- Creates tight, compact clusters
- Minimizes within-cluster variance
- Single → chain effect (bad)
- Complete → too strict
- Average → decent but not best
- Ward → best for recommendation systems
