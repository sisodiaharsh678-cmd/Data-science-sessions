2. Rewriting a Vague Food Prompt

Suggest 7 easy vegetarian dinner ideas for a college student living 
in a hostel in Ahmedabad. Assume limited cooking equipment (just an 
induction cooktop or electric kettle, no full kitchen), a tight 
budget, and ingredients easily available in local Gujarati grocery 
stores. Include a mix of North Indian and Gujarati-style dishes, 
and mention approximate prep time for each.


1. Gujarati Moong Dal Khichdi 🥣

Put rice + moong dal + turmeric + salt + water in one pot and cook until soft. Add a little jeera-ghee/oil tempering if you have time. Gujarati khichdi traditionally uses rice and moong dal with simple seasoning and is commonly served as khichdi-kadhi.

Hostel advantage: One pot, cheap, filling and very little cleaning.

2. Besan Chilla

Mix besan + water + salt + chilli + onion/tomato into a batter and make pancake-like chillas on your induction pan. Serve with curd.

Time: ~15 minutes.
Budget: Very cheap, and besan is easy to find locally.

3. Veg Poha

Soak/rinse poha, then sauté onion + potato + peanuts + spices, add poha and finish with lemon.

Time: ~15 minutes.
Great for nights when you don't want to cook a full meal.

4. Dal Tadka + Rice

Cook rice and moong/toor dal. For the tadka, heat a little oil and add jeera, chilli, turmeric and tomato.

Time: ~25–30 minutes.

5. Sev Tameta 🍅

Cook chopped tomatoes with salt, turmeric, chilli and a little oil until soft. Add sev only at the end so it doesn't become completely soggy. Eat with bread or ready-made rotis.

Time: ~15 minutes.

6. Vegetable Upma

Roast rava, then cook it with water, onion, tomato and whatever cheap vegetables you have.

Time: ~15–20 minutes.

7. Soya-Paneer Masala 💪

Boil/soak soya chunks, then cook them with onion, tomato and basic masala. Add some paneer if your budget allows. Eat with rice.

Time: ~25 minutes.
