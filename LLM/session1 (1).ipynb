# LLM - Large Language Model

```python
# Large = Trained on huge amounts of text
# Language = Works with human language (specifically on text (English, Hindi, code, etc))
# Model = Mathematical system that predicts and generates text
```

A Large Language Model (LLM) is an AI system trained on massive amounts of text. It learns patterns in language - grammar, facts, reasoning, style and uses that knowledge to understand and generate human-like text.

```python
# How LLM Understands Text

# Sentence > LLM converts words into numbers > Computer only understands numbers > Transformer finds relationships between words
```

#### What is Transformer?


```python
# Transformer is the brain behind modern LLMs.

# Before Transformers:
# Word 1 → Word 2 → Word 3 → Word 4
# Read one by one.

# After Transformer:
# Word 1 <-> Word 2
# Word 1 <-> Word 3
# Word 1 <-> Word 4
# Looks at all words together

# This makes it much faster and smarter
```

Next Word Prediction

```python
text = "I drink tea in the"
words = text.split()
print("Input Sentence:", text)

predictions = {"the": "morning"}
last_word = words[-1]
print("Model Prediction:",predictions[last_word])

# This program knows 1 pattern
# LLM knows billions of patterns
```

```python
# Transformer Architecture

# Input Sentence > Tokenization (break into pieces) > Embeddings > Self-Attention (Find important words) >
# > Feed Forward Network(Learns patterns and relationships) > Output Word(Predicts the next token)
```

Mini AI Chatbot

```python
while True:
    user = input("You: ")

    if user.lower() == "hello":
        print("Bot: Hi there!")

    elif user.lower() == "how are you":
        print("Bot: I am fine.")

    elif user.lower() == "bye":
        print("Bot: Goodbye!")
        break

    else:
        print("Bot: I don't know.")
```

```python
# Normal chatbot vs LLM - what's the difference?
# Traditional chatbots follow fixed scripts. 
# LLMs generate responses dynamically, handling conversations they've never seen before.

# Traditional chatbot         VS          LLM-powered assistant

# Rule-based (if/else logic)              Probabilistic generation(Learns patterns)
# Pre-written responses only              Creates new responses each time
# Fails on unexpected input               Handles any input gracefully
# Can't reason or infer                   Can reason, summarise, explain
# Easy to build, limited                  Requires huge compute to build
# Example: Bank phone menu                Example: ChatGPT, Claude
```

```python
# Simple text prediction (no API needed)

import random

# A very simple "language model" — just a dictionary
simple_model = {
    "The sky is":   ["blue", "cloudy", "clear", "beautiful"],
    "I love to":    ["code", "learn", "eat", "sleep"],
    "Python is":    ["fun", "powerful", "easy", "great"],
}

def predict_next(sentence):
    if sentence in simple_model:
        # Pick a random option (real LLMs use probabilities)
        return random.choice(simple_model[sentence])
    return "I don't know yet!"

# Try it
print(predict_next("The sky is"))
print(predict_next("Python is"))
print(predict_next("I love to"))
```

### GPT vs BERT vs T5

```python
# one transformer, three different ideas
```

```python
# GPT Architecture

# Input Text > Embedding > Decoder Layer > Decoder Layer > Decoder Layer > Next Word Prediction
```

```python
# GPT — Decoder-only transformer
# GPT uses only the decoder part of the original transformer. 
# It reads text from left to right and predicts the next token. 
# This is called causal (or autoregressive) language modelling.
```

```python
# Training objective: next-token prediction
# During training, GPT sees "The cat sat on" and predicts "mat". 
# Then "The cat sat on mat" and predicts the next word. 
# This is called CLM (Causal Language Modelling). The model sees billions of such sequences.
```

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="gpt2"
)

result = generator(
    "Artificial Intelligence is",
    max_length=50
)

print(result[0]["generated_text"])
```

```python
# BERT — Encoder-only transformer
# BERT uses only the encoder part. 
# It reads the entire sentence in both directions at once — left AND right context. 
# This is called bidirectional attention.
```

```python
# Also trained with Next Sentence Prediction (NSP)
# BERT also learns to predict whether two sentences naturally follow each other. 
# This helps it understand relationships between sentences — useful for question answering.
```

```python
# T5 — Encoder-Decoder transformer
# T5 uses both the encoder and decoder. The encoder reads and understands the input. 
# The decoder generates the output. Every task is framed as "text in → text out".
```

```python
#  GPT — Text generation
from transformers import pipeline

# Load GPT-2 (free, open source)
generator = pipeline("text-generation", model="gpt2")

# Generate text from a prompt
result = generator(
    "Once upon a time in India,",
    max_length=60,          # Stop after 60 tokens
    num_return_sequences=1, # Generate 1 version
    temperature=0.8         # Creativity: 0=safe, 1=creative
)

print(result[0]["generated_text"])
```

```python
# BERT — Sentiment analysis
from transformers import pipeline

# Load sentiment model (fine-tuned BERT)
classifier = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# Classify sentiment of reviews
reviews = [
    "This movie was absolutely amazing!",
    "I hated every minute of it.",
    "It was okay, nothing special."
]

for review in reviews:
    result = classifier(review)[0]
    print(f"Text: {review}")
    print(f"Sentiment: {result['label']}, Score: {result['score']:.2f}")
    print()
```

```python
# T5 — Summarisation

from transformers import pipeline

# Load T5 summarisation pipeline
summarizer = pipeline("summarization", model="t5-small")

# Long text to summarise
long_text = """
Artificial intelligence is a branch of computer science that
aims to create machines that can perform tasks that normally
require human intelligence. These tasks include understanding
natural language, recognizing patterns, solving problems, and
making decisions. AI is used in many fields today, including
healthcare, finance, education, and transportation. Machine
learning, a subset of AI, allows computers to learn from data
without being explicitly programmed.
"""

summary = summarizer(
    long_text,
    max_length=60,   # Maximum summary length
    min_length=20,   # Minimum summary length
    do_sample=False  # Deterministic output
)

print("Summary:", summary[0]["summary_text"])
```

```python
# BERT — Extractive question answering
from transformers import pipeline

# BERT finds the answer WITHIN the context (extractive)
qa = pipeline(
    "question-answering",
    model="distilbert-base-uncased-distilled-squad"
)

context = """
India is a country in South Asia. It is the seventh-largest
country in the world by area and the most populous country.
New Delhi is the capital of India. India achieved independence
from British rule on 15 August 1947.
"""

questions = [
    "What is the capital of India?",
    "When did India gain independence?",
    "How large is India by area ranking?"
]

for question in questions:
    result = qa(question=question, context=context)
    print(f"Q: {question}")
    print(f"A: {result['answer']} (confidence: {result['score']:.2f})")
    print()
```

```python
# T5 — Multiple tasks with one model
from transformers import T5ForConditionalGeneration, T5Tokenizer

# Load T5 model and tokenizer
model = T5ForConditionalGeneration.from_pretrained("t5-small")
tokenizer = T5Tokenizer.from_pretrained("t5-small")

def t5_task(task_prefix, input_text):
    """T5 uses text prefixes to know the task"""
    # Format: "task: input_text"
    full_input = f"{task_prefix}: {input_text}"

    # Tokenize (convert text to numbers)
    inputs = tokenizer(full_input, return_tensors="pt",
                       max_length=512, truncation=True)

    # Generate output
    outputs = model.generate(
        inputs["input_ids"],
        max_length=150,
        num_beams=4,      # Search 4 options, pick best
        early_stopping=True
    )

    # Decode (convert numbers back to text)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

# Same model, different tasks!
print("SUMMARISE:")
print(t5_task("summarize",
    "The Eiffel Tower is a wrought-iron lattice tower in Paris. "
    "It was constructed from 1887 to 1889 as the entrance arch "
    "for the 1889 World's Fair."))

print("\nTRANSLATE:")
print(t5_task("translate English to French", "Hello, how are you?"))

print("\nQUESTION:")
print(t5_task("question: What is Python? context",
    "Python is a high-level programming language known for "
    "its simple syntax and readability."))
```
