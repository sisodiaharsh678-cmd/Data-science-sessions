```python
#Task 1: GPT-2 caption generation 

from transformers import pipeline 

generator = pipeline("text-generation" , model= 'gpt2')

prompt = "sunset vibes at the beach"

result  = generator(prompt , max_length = 30 , num_return_sequences=1)

print(result[0]['generated_text'])
```

```python
#Task 2: BERT sentiment classification 
from transformers import pipeline

classifier = pipeline('sentiment-analysis')

review = "The food was not bad and the delivery was faster and well."
result = classifier(review)

print(result)
```

```python
#T5 

import os
os.environ["TOKENIZERS_PARALLELISM"] = "false"

from transformers import T5Tokenizer, T5ForConditionalGeneration

tokenizer = T5Tokenizer.from_pretrained('Vamsi/T5_Paraphrase_Paws')
model = T5ForConditionalGeneration.from_pretrained('Vamsi/T5_Paraphrase_Paws')

input_text = "paraphrase: where are you when u come . </s>"

encoding = tokenizer.encode_plus(
    input_text,
    padding='max_length',
    max_length=64,
    return_tensors="pt",
    truncation=True,
)

output_ids = model.generate(
    input_ids=encoding["input_ids"],
    attention_mask=encoding["attention_mask"],
    max_length=64,
    num_beams=5,
    num_return_sequences=5,
    do_sample=True,
    temperature=0.8,
    top_k=120,
    top_p=0.95,
    early_stopping=True,
)

for output in output_ids:
    print(tokenizer.decode(output, skip_special_tokens=True))
```

#Task 5: Which model fits each scenario

1. Generating a playlist description for Spotify → GPT
This is a pure text-generation task — creating new descriptive text from scratch — which is exactly GPT's decoder-only strength.
2. Classifying spam messages on WhatsApp → BERT
This is a classification task on existing text (spam vs. not spam), which needs deep bidirectional understanding of the message — BERT's core strength, not generation.
3. Summarizing a cricket match report for an IPL app → T5
This is a text-to-text transformation — taking a long report in and producing a shorter summary out — which is precisely what T5's encoder-decoder architecture is built for.
