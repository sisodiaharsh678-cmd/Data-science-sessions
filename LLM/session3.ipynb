```python
import requests

response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "llama2",
        "prompt": "Suggest 5 new features for a food delivery app like Zomato that use AI.",
        "stream": False
    }
)

print(response.json()["response"])
```
