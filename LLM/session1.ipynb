Question 2: How does an LLM generate a reply?

When you send a message, the LLM first breaks your text into smaller pieces called tokens, which it converts into numbers it can process. It then uses its training — patterns learned from massive amounts of text during development — to understand the context and intent behind your question. Based on this understanding, it predicts the most likely next word, one token at a time, building up a full response step by step. Each new word is chosen based on everything that came before it in the conversation, which is why the reply stays coherent and relevant to what you asked. Finally, the generated tokens are converted back into readable text and sent to you as the reply


Question 3: Rule-based chatbot vs. LLM-powered chatbot
How it handles queries: A rule-based chatbot (like an old IRCTC bot) matches your message against a fixed set of pre-written keywords or commands. An LLM-powered chatbot understands the actual meaning and intent behind your message, even if phrased differently.

Language flexibility: A rule-based chatbot is very rigid and breaks if you don't use the exact expected phrasing. An LLM-powered chatbot handles natural, conversational language, typos, and varied phrasing.

Learning capability: A rule-based chatbot cannot learn or improve on its own and needs manual updates by developers. An LLM-powered chatbot learned from massive training data and can adapt within a conversation using context.

Handling unexpected questions: A rule-based chatbot fails or gives a generic "I don't understand" response. An LLM-powered chatbot attempts a relevant, coherent answer even for unfamiliar or complex questions.


Question 4: LLM improving a daily app

In Zomato, an LLM could power a smart food-recommendation assistant that understands natural requests like "suggest something light and spicy under ₹200 near me" instead of forcing users to manually filter through cuisine and price options. It could also generate personalized dish descriptions based on a user's past orders and preferences, making suggestions feel more relevant. Additionally, an LLM-powered support chatbot could handle order-related questions like delays or refunds more naturally, understanding varied complaint phrasing instead of relying on fixed menu options. This would make the overall experience faster and less frustrating for users.


Question 5: LLM advantages/limitations — your own ChatGPT interaction

Prompt used: "What are two advantages and two limitations of using LLMs in business applications?"


Two advantages of LLMs in business

Increased productivity – LLMs can automate tasks such as writing emails, summarizing documents, generating reports, and answering customer questions, saving employees time.

Flexible applications – The same LLM can be used for customer support, content creation, data analysis, coding assistance, and knowledge retrieval.

Two limitations of LLMs

Hallucinations / inaccurate information – LLMs can produce confident but incorrect or fabricated answers, so important business outputs need human verification.

Privacy, security, and cost concerns – Using LLMs with sensitive business data requires proper security and governance, and operating them at scale can become expensive.
