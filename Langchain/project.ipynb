```python
from langchain_google_genai import GoogleGenerativeAI

llm = GoogleGenerativeAI(model="gemini-3-flash-preview", temperature=0.6)
name = llm.invoke("I want to open a restaurant for Indian food. Suggest a fancy name for this.")
print(name)
```

Prompt

```python
from langchain_core.prompts import PromptTemplate

prompt_template_name = PromptTemplate(
    input_variables = ["cuisine"],
    template = "I want to open a restaurant for {cuisine} food . suggest a fency name for this."

)

prompt_template_name.format(cuisine="Italian")
```

Chain

```python
chain = prompt_template_name | llm
response = chain.invoke({"cuisine": "mexican"})
print(response)
```
