```python
from langchain_core.tools import tool 

@tool 
def get_weather(city:str)-> str:
    """Get the current weather for a given city."""
    return f"The weather in {city} is sunny, 28°C."


result =get_weather.invoke({"city":"surat"})
print(result)


print(get_weather.name)          # "get_weather"  — from the function name
print(get_weather.description)   # "Get the current weather for a given city."  — from the docstring
print(get_weather.args)          # {'city': {'title': 'City', 'type': 'string'}}  — from type hints
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI



from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-3-flash-preview")
model_with_tools = model.bind_tools([get_weather])

response = model_with_tools.invoke("What's the weather in Mumbai?")
print(response.tool_calls)
```
